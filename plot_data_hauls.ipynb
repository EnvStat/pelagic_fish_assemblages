```python
P = artist.Panel()
P.Y.k=50
P.X.k=80
for i, c in enumerate(data.country.unique()):
    P.text(i, 1, str(c), pos='..', z=1, font=20)
    for j, year in enumerate(sorted(data.year.unique())):
        x = sum((data.year == year) & (data.country == c))
        if x > 0:
            P.circle(i, -j, r=x**0.5 * 3.5, col='o')
            P.text(i, -j, str(x), pos='..', font=20)
            
for j, year in enumerate(sorted(data.year.unique())):
    P.text(-1, -j, str(year), pos='<.', z=1, font=20)

P.top = 'hauls per country'
P.show()
```

```python
P = make_map()
P.background('_w')

P.points(data.x_lon, data.y_lat, col='k', p=7)

P.title = 'data points'
P.show()
```

```python
P = make_map()
P.background('__w')
colors = [artist.Color(x) for x in data.country]
P.points(data.x_lon, data.y_lat, colors=colors, p=7)

countries = list(data.country.unique())
for c in countries:
    subset = data[data.country == c]
    x, y = subset.x_lon.mean(), subset.y_lat.mean()
    P.circle(x, y, r=70, col='__w')
    P.text(x, y, c, font=70, pos='..')
P.show()
```

```python
def year_to_x(year):
    return (year-2)%5 * 20
def year_to_y(year):
    return (year-2)//5 * -13
```

```python
P = artist.Panel(2000, 1600)

colors = [artist.Color(x) for x in data.country]
P.points(data.x_lon + year_to_x(data.year), data.y_lat + year_to_y(data.year), colors=colors, p=11)
for y in data.year.unique():
    P.text(year_to_x(y)+13, year_to_y(y)+60, str(y))


P.show()
```

```python
from useful import rnorm

def plot_cov(cov, label, C=(1,)):
    P = make_map()
    P.C.fit(1)
    if log:
        P.C.fit(*C)
    P.background('__w')

    P.points(data.x_lon+rnorm(0, 0.1, len(data)), data.y_lat+rnorm(0, 0.1, len(data)), colors=data[cov], p=9)
    print(data[cov].min(), data[cov].max())
    P.C.grad = 'heat'
    return (P/P.color_axis('v', 600, 20, label)).show()


plot_cov('depth', 'Depth, m')
```
